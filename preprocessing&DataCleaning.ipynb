```python
import pandas as pd
import re
from pathlib import Path

def sample_tsv(path, n=5000, chunksize=100_000, random_state=42):
    """
    Sample approximately n rows from a large TSV without loading
    the complete file into memory.
    """
    samples = []

    for chunk in pd.read_csv(
        path,
        sep="\t",
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ],
        chunksize=chunksize
    ):
        take = min(len(chunk), max(1, n // 10))

        samples.append(
            chunk.sample(
                n=take,
                random_state=random_state
            )
        )

        if sum(len(x) for x in samples) >= n:
            break

    result = pd.concat(samples, ignore_index=True)

    if len(result) > n:
        result = result.sample(
            n=n,
            random_state=random_state
        ).reset_index(drop=True)

    return result
```

#  imports + normalize_unicode()

```python
%%writefile -a /kaggle/working/src/preprocess.py

import re
import unicodedata


def normalize_unicode(text):
    """
    Normalize Unicode text while preserving non-ASCII characters.

    Handles:
    - None / missing values
    - Unicode compatibility normalization (NFKC)
    - Leading/trailing whitespace
    - Repeated whitespace
    """

    if text is None:
        return ""

    text = str(text)

    # Normalize Unicode while preserving characters such as
    # Hindi, French accents, etc.
    text = unicodedata.normalize("NFKC", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

# normalize_business_name()

```python
%%writefile -a /kaggle/working/src/preprocess.py

def normalize_business_name(text):
    """
    Normalize business names for entity matching.

    Keeps Unicode characters intact while:
    - normalizing Unicode
    - converting to lowercase
    - expanding common business abbreviations
    - converting '&' to 'and'
    - removing punctuation
    - normalizing whitespace
    """

    text = normalize_unicode(text)

    if not text:
        return ""

    # Lowercase while preserving Unicode
    text = text.lower()

    # Normalize common business abbreviations
    replacements = {
        r"\bpvt\.?\b": "private",
        r"\bltd\.?\b": "limited",
        r"\bcorp\.?\b": "corporation",
    }

    for pattern, replacement in replacements.items():
        text = re.sub(pattern, replacement, text)

    # '&' → 'and'
    text = text.replace("&", " and ")

    # Remove punctuation while preserving Unicode letters,
    # numbers, and combining marks.
    cleaned = []

    for char in text:
        category = unicodedata.category(char)

        if category.startswith("P"):
            cleaned.append(" ")
        else:
            cleaned.append(char)

    text = "".join(cleaned)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

# normalize_business_address()

```python
%%writefile -a /kaggle/working/src/preprocess.py

def normalize_business_address(text):
    """
    Normalize business addresses for entity matching.

    Keeps useful address information such as:
    - house numbers
    - street numbers
    - postal codes
    - Unicode characters

    Normalizes:
    - Unicode
    - lowercase
    - common address abbreviations
    - '&' to 'and'
    - punctuation
    - whitespace
    """

    text = normalize_unicode(text)

    if not text:
        return ""

    # Lowercase while preserving Unicode
    text = text.lower()

    # Common address abbreviations.
    # These are applied token-wise so that words containing
    # 'st' or 'rd' are not accidentally modified.
    replacements = {
        r"\bst\.?\b": "street",
        r"\brd\.?\b": "road",
    }

    for pattern, replacement in replacements.items():
        text = re.sub(pattern, replacement, text)

    # '&' → 'and'
    text = text.replace("&", " and ")

    # Remove punctuation while preserving Unicode letters,
    # numbers, and combining marks.
    cleaned = []

    for char in text:
        category = unicodedata.category(char)

        if category.startswith("P"):
            cleaned.append(" ")
        else:
            cleaned.append(char)

    text = "".join(cleaned)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

# extract_address_features()

```python
%%writefile -a /kaggle/working/src/preprocess.py

def extract_address_features(text):
    """
    Extract country-agnostic numeric information from an address.

    Returns:
        A dictionary containing:
        - numbers: numeric sequences found in the address

    No country-specific postal-code assumptions are made.
    """

    text = normalize_unicode(text)

    if not text:
        return {
            "numbers": []
        }

    # Extract standalone numeric sequences.
    # Examples:
    # "2100 Cameron Drive" -> ["2100"]
    # "Unit 4 Building 12" -> ["4", "12"]
    numbers = re.findall(r"\d+", text)

    return {
        "numbers": numbers
    }
```

```python
%%writefile -a /kaggle/working/src/preprocess.py

import pandas as pd


def preprocess_dataframe(df):
    """
    Apply preprocessing to a source dataframe.

    Original columns are preserved.
    New columns:
        - business_name_clean
        - business_address_clean
        - address_numbers
    """

    df = df.copy()

    # Normalize business names
    df["business_name_clean"] = (
        df["business_name"]
        .apply(normalize_business_name)
    )

    # Normalize business addresses
    df["business_address_clean"] = (
        df["business_address"]
        .apply(normalize_business_address)
    )

    # Extract numeric address information
    df["address_numbers"] = (
        df["business_address"]
        .apply(lambda x: extract_address_features(x)["numbers"])
    )

    return df
```

```python
import pandas as pd
import re
from collections import Counter
from pathlib import Path


# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/kaggle/input/datasets/lakshaygauniyal/"
    "amazon-ml-2026-dataset/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource/dataset"
)

FILES = {
    "train_source1": BASE / "train/train_source1.tsv",
    "train_source2": BASE / "train/train_source2.tsv",
    "train_source3": BASE / "train/train_source3.tsv",
    "test_source1": BASE / "test/test_source1.tsv",
    "test_source2": BASE / "test/test_source2.tsv",
    "test_source3": BASE / "test/test_source3.tsv",
}


# ============================================================
# SETTINGS
# ============================================================

CHUNK_SIZE = 50_000

FAKE_NULLS = {
    "null",
    "none",
    "na",
    "n/a",
    "nan",
    "nil",
    "unknown"
}


# Indic Unicode ranges
INDIC_RANGES = [
    (0x0900, 0x097F),  # Devanagari
    (0x0980, 0x09FF),  # Bengali
    (0x0A00, 0x0A7F),  # Gurmukhi
    (0x0A80, 0x0AFF),  # Gujarati
    (0x0B00, 0x0B7F),  # Odia
    (0x0B80, 0x0BFF),  # Tamil
    (0x0C00, 0x0C7F),  # Telugu
    (0x0C80, 0x0CFF),  # Kannada
    (0x0D00, 0x0D7F),  # Malayalam
]


def contains_indic(text):
    if pd.isna(text):
        return False

    for char in str(text):
        code = ord(char)

        for start, end in INDIC_RANGES:
            if start <= code <= end:
                return True

    return False


def extract_words(text):
    """
    Unicode-aware word extraction.
    """
    return re.findall(
        r"\b[\wÀ-ÖØ-öø-ÿ]+\b",
        str(text).lower(),
        flags=re.UNICODE
    )


# ============================================================
# MAIN EDA FUNCTION
# ============================================================

def run_eda(filepath, source_name):

    print("\n" + "=" * 80)
    print(f"PROCESSING: {source_name}")
    print("=" * 80)

    # Counters
    france_name_words = Counter()
    france_address_words = Counter()

    address_two_letter_tokens = Counter()
    fake_null_counter = Counter()

    # Row statistics
    total_rows = 0

    country_total = Counter()
    country_missing_address = Counter()

    india_total = 0
    india_indic_rows = 0

    # --------------------------------------------------------
    # READ FILE IN CHUNKS
    # --------------------------------------------------------

    for chunk_number, chunk in enumerate(
        pd.read_csv(
            filepath,
            sep="\t",
            usecols=[
                "business_name",
                "business_address",
                "country"
            ],
            chunksize=CHUNK_SIZE
        ),
        start=1
    ):

        total_rows += len(chunk)

        # ----------------------------------------------------
        # COUNTRY
        # ----------------------------------------------------

        country_series = (
            chunk["country"]
            .fillna("")
            .astype(str)
            .str.strip()
            .str.lower()
        )

        # ----------------------------------------------------
        # MISSING ADDRESS
        # ----------------------------------------------------

        missing_address = chunk["business_address"].isna()

        for country, count in country_series.value_counts().items():

            country_total[country] += count

            country_missing_address[country] += (
                missing_address[country_series == country].sum()
            )

        # ----------------------------------------------------
        # INDIA + INDIC SCRIPT
        # ----------------------------------------------------

        india_mask = country_series == "india"

        india_rows = chunk[india_mask]

        india_total += len(india_rows)

        if len(india_rows) > 0:

            name_indic = india_rows["business_name"].apply(
                contains_indic
            )

            address_indic = india_rows["business_address"].apply(
                contains_indic
            )

            indic_any = name_indic | address_indic

            india_indic_rows += indic_any.sum()

        # ----------------------------------------------------
        # FRANCE VOCABULARY
        # ----------------------------------------------------

        france_mask = country_series == "france"

        france_rows = chunk[france_mask]

        for text in france_rows["business_name"].dropna():
            france_name_words.update(
                extract_words(text)
            )

        for text in france_rows["business_address"].dropna():
            france_address_words.update(
                extract_words(text)
            )

        # ----------------------------------------------------
        # ADDRESS ABBREVIATIONS + FAKE NULLS
        # ----------------------------------------------------

        for text in chunk["business_address"].dropna():

            text_lower = str(text).lower()

            # Two-letter alphabetic tokens
            two_letter_tokens = re.findall(
                r"\b[a-zA-Z]{2}\b",
                text_lower
            )

            address_two_letter_tokens.update(
                two_letter_tokens
            )

            # Fake null tokens
            tokens = re.findall(
                r"\b[\w/.-]+\b",
                text_lower
            )

            for token in tokens:

                if token in FAKE_NULLS:
                    fake_null_counter[token] += 1

        # ----------------------------------------------------
        # PROGRESS
        # ----------------------------------------------------

        if chunk_number % 5 == 0:
            print(
                f"Processed {total_rows:,} rows..."
            )

    # ========================================================
    # RESULTS
    # ========================================================

    print("\n" + "-" * 80)
    print("BASIC INFORMATION")
    print("-" * 80)

    print("Total rows:", f"{total_rows:,}")

    # --------------------------------------------------------
    # 1. FRANCE VOCABULARY
    # --------------------------------------------------------

    print("\n" + "-" * 80)
    print("1. FRANCE VOCABULARY — BUSINESS NAMES")
    print("-" * 80)

    print(
        france_name_words.most_common(30)
    )

    print("\n" + "-" * 80)
    print("1. FRANCE VOCABULARY — ADDRESSES")
    print("-" * 80)

    print(
        france_address_words.most_common(30)
    )

    # --------------------------------------------------------
    # 2. ADDRESS ABBREVIATIONS
    # --------------------------------------------------------

    print("\n" + "-" * 80)
    print("2. COMMON TWO-LETTER ADDRESS TOKENS")
    print("-" * 80)

    print(
        address_two_letter_tokens.most_common(50)
    )

    # --------------------------------------------------------
    # 2B. FAKE NULLS
    # --------------------------------------------------------

    print("\n" + "-" * 80)
    print("2B. FAKE NULL WORDS IN ADDRESSES")
    print("-" * 80)

    print(
        dict(fake_null_counter)
    )

    # --------------------------------------------------------
    # 3. INDIC SCRIPT
    # --------------------------------------------------------

    print("\n" + "-" * 80)
    print("3. INDIC SCRIPT — INDIA")
    print("-" * 80)

    indic_percentage = (
        india_indic_rows / india_total * 100
        if india_total > 0
        else 0
    )

    print(
        f"Indic rows: {india_indic_rows:,} / "
        f"{india_total:,}"
    )

    print(
        f"Percentage: {indic_percentage:.2f}%"
    )

    # --------------------------------------------------------
    # 4. MISSING ADDRESS
    # --------------------------------------------------------

    print("\n" + "-" * 80)
    print("4. MISSING ADDRESS BY COUNTRY")
    print("-" * 80)

    for country in ["us", "india", "france"]:

        total = country_total[country]
        missing = country_missing_address[country]

        percentage = (
            missing / total * 100
            if total > 0
            else 0
        )

        print(
            f"{country.upper():8} "
            f"{missing:,} / {total:,} "
            f"= {percentage:.2f}%"
        )

    # Return results as a dictionary too
    return {
        "total_rows": total_rows,
        "france_name_top30": france_name_words.most_common(30),
        "france_address_top30": france_address_words.most_common(30),
        "two_letter_address_tokens": address_two_letter_tokens.most_common(50),
        "fake_nulls": dict(fake_null_counter),
        "india_total": india_total,
        "india_indic_rows": india_indic_rows,
        "india_indic_percentage": indic_percentage,
        "country_total": dict(country_total),
        "country_missing_address": dict(country_missing_address)
    }
```

```python
# ============================================================
# CREATE MATCHED RECORDS INSIDE DUCKDB
# ============================================================

con.execute("DROP TABLE IF EXISTS matched_records")

query = f"""
CREATE TEMP TABLE matched_records AS

-- ==========================================================
-- S1 <-> S2
-- ==========================================================

SELECT
    gt.source1_entity_id AS s1_id,

    s1.business_name    AS s1_name,
    s1.business_address AS s1_address,

    TRIM(matched_id)    AS matched_id,
    'S2'                AS matched_source,

    s2.business_name    AS matched_name,
    s2.business_address AS matched_address

FROM read_csv_auto(
    '{GROUND_TRUTH}',
    delim='\\t'
) gt

JOIN read_csv_auto(
    '{SOURCE1}',
    delim='\\t'
) s1
    ON s1.entity_id = gt.source1_entity_id

CROSS JOIN UNNEST(
    STRING_SPLIT(gt.matched_entity_ids, ',')
) AS u(matched_id)

JOIN read_csv_auto(
    '{SOURCE2}',
    delim='\\t'
) s2
    ON s2.entity_id = TRIM(matched_id)

WHERE TRIM(matched_id) LIKE 'S2-%'


UNION ALL


-- ==========================================================
-- S1 <-> S3
-- ==========================================================

SELECT
    gt.source1_entity_id AS s1_id,

    s1.business_name    AS s1_name,
    s1.business_address AS s1_address,

    TRIM(matched_id)    AS matched_id,
    'S3'                AS matched_source,

    s3.business_name    AS matched_name,
    s3.business_address AS matched_address

FROM read_csv_auto(
    '{GROUND_TRUTH}',
    delim='\\t'
) gt

JOIN read_csv_auto(
    '{SOURCE1}',
    delim='\\t'
) s1
    ON s1.entity_id = gt.source1_entity_id

CROSS JOIN UNNEST(
    STRING_SPLIT(gt.matched_entity_ids, ',')
) AS u(matched_id)

JOIN read_csv_auto(
    '{SOURCE3}',
    delim='\\t'
) s3
    ON s3.entity_id = TRIM(matched_id)

WHERE TRIM(matched_id) LIKE 'S3-%'
"""

print("Building matched_records...")
con.execute(query)

print("Done.")
```

```python
# ============================================================
# STEP 5A
# DISCOVER FREQUENT TOKEN DIFFERENCES
# ============================================================

import re
from collections import Counter


def tokenize(text):
    if text is None:
        return []

    text = str(text).lower()

    # Keep Unicode letters/numbers
    tokens = re.findall(r"[^\W_]+", text, flags=re.UNICODE)

    return tokens


# Take a manageable sample first
sample_pairs = con.execute("""
    SELECT
        s1_name,
        matched_name,
        s1_address,
        matched_address,
        matched_source
    FROM matched_records
    USING SAMPLE 100000 ROWS
""").fetchdf()

print("Sample size:", len(sample_pairs))


# ------------------------------------------------------------
# Collect token differences
# ------------------------------------------------------------

name_additions = Counter()
name_removals = Counter()

address_additions = Counter()
address_removals = Counter()


for row in sample_pairs.itertuples(index=False):

    # -------------------------
    # BUSINESS NAME
    # -------------------------

    s1_tokens = set(tokenize(row.s1_name))
    matched_tokens = set(tokenize(row.matched_name))

    additions = matched_tokens - s1_tokens
    removals = s1_tokens - matched_tokens

    for token in additions:
        name_additions[token] += 1

    for token in removals:
        name_removals[token] += 1


    # -------------------------
    # ADDRESS
    # -------------------------

    s1_tokens = set(tokenize(row.s1_address))
    matched_tokens = set(tokenize(row.matched_address))

    additions = matched_tokens - s1_tokens
    removals = s1_tokens - matched_tokens

    for token in additions:
        address_additions[token] += 1

    for token in removals:
        address_removals[token] += 1


print("\nTop name additions:")
print(name_additions.most_common(30))

print("\nTop name removals:")
print(name_removals.most_common(30))

print("\nTop address additions:")
print(address_additions.most_common(30))

print("\nTop address removals:")
print(address_removals.most_common(30))
```

```python
import re
import unicodedata
from collections import Counter


# ============================================================
# TOKENIZATION FOR ABBREVIATION DISCOVERY
# ============================================================

def comparison_tokens(text):
    """
    Normalize only for discovering token substitutions.

    IMPORTANT:
    This is NOT our final preprocessing normalization.
    """

    if text is None:
        return []

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Lowercase
    text = text.lower()

    # Remove punctuation WITHOUT converting it to spaces.
    # Example:
    #   L.L.C. -> llc
    #   St.     -> st
    text = "".join(
        char
        for char in text
        if not unicodedata.category(char).startswith("P")
    )

    # Unicode-aware tokenization
    return re.findall(r"[^\W_]+", text, flags=re.UNICODE)


# ============================================================
# DISCOVER ONE-TOKEN SUBSTITUTIONS
# ============================================================

name_changes = Counter()
address_changes = Counter()


for row in sample_pairs.itertuples(index=False):

    # --------------------------------------------------------
    # BUSINESS NAME
    # --------------------------------------------------------

    a = comparison_tokens(row.s1_name)
    b = comparison_tokens(row.matched_name)

    if len(a) == len(b):

        differences = [
            (x, y)
            for x, y in zip(a, b)
            if x != y
        ]

        # EXACTLY ONE token changed
        if len(differences) == 1:

            old_token, new_token = differences[0]

            name_changes[(old_token, new_token)] += 1


    # --------------------------------------------------------
    # ADDRESS
    # --------------------------------------------------------

    a = comparison_tokens(row.s1_address)
    b = comparison_tokens(row.matched_address)

    if len(a) == len(b):

        differences = [
            (x, y)
            for x, y in zip(a, b)
            if x != y
        ]

        # EXACTLY ONE token changed
        if len(differences) == 1:

            old_token, new_token = differences[0]

            address_changes[(old_token, new_token)] += 1


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("NAME TOKEN SUBSTITUTIONS")
print("=" * 60)

for (old, new), count in name_changes.most_common(50):
    print(f"{old:25} -> {new:25} : {count}")


print("\nADDRESS TOKEN SUBSTITUTIONS")
print("=" * 60)

for (old, new), count in address_changes.most_common(50):
    print(f"{old:25} -> {new:25} : {count}")
```

# Confidence calculation

```python
from collections import defaultdict, Counter

# ============================================================
# BUILD TOKEN TRANSITION COUNTS
#
# For every comparable pair where ONLY ONE TOKEN differs:
#
#       old_token -> new_token
#
# Also count unchanged tokens:
#
#       old_token -> same_token
# ============================================================


def collect_token_statistics(rows, text_col_1, text_col_2):

    transitions = defaultdict(Counter)

    for row in rows.itertuples(index=False):

        a = comparison_tokens(getattr(row, text_col_1))
        b = comparison_tokens(getattr(row, text_col_2))

        # Only compare records with the same number of tokens
        if len(a) != len(b):
            continue

        differences = [
            (x, y)
            for x, y in zip(a, b)
            if x != y
        ]

        # We only want ONE-token differences
        if len(differences) != 1:
            continue

        old_token, new_token = differences[0]

        transitions[old_token][new_token] += 1

    return transitions


# ============================================================
# NAME
# ============================================================

name_transitions = collect_token_statistics(
    sample_pairs,
    "s1_name",
    "matched_name"
)


# ============================================================
# ADDRESS
# ============================================================

address_transitions = collect_token_statistics(
    sample_pairs,
    "s1_address",
    "matched_address"
)


# ============================================================
# SHOW CANDIDATES WITH CONFIDENCE
# ============================================================

def show_candidates(transitions, min_count=20):

    candidates = []

    for old_token, counter in transitions.items():

        total = sum(counter.values())

        new_token, count = counter.most_common(1)[0]

        confidence = count / total

        candidates.append(
            (
                old_token,
                new_token,
                count,
                total,
                confidence
            )
        )

    candidates.sort(
        key=lambda x: (x[4], x[2]),
        reverse=True
    )

    for old, new, count, total, confidence in candidates:

        if count >= min_count:

            print(
                f"{old:20} -> "
                f"{new:20} | "
                f"count={count:5} | "
                f"total={total:5} | "
                f"confidence={confidence:.3f}"
            )


print("=" * 80)
print("BUSINESS NAME CANDIDATES")
print("=" * 80)

show_candidates(name_transitions)


print("\n" + "=" * 80)
print("ADDRESS CANDIDATES")
print("=" * 80)

show_candidates(address_transitions)
```

```python
# ============================================================
# FILTER ADDRESS CANDIDATES
# Separate likely state mappings from actual abbreviations
# ============================================================

US_STATES = {
    "alabama", "alaska", "arizona", "arkansas", "california",
    "colorado", "connecticut", "delaware", "florida", "georgia",
    "hawaii", "idaho", "illinois", "indiana", "iowa",
    "kansas", "kentucky", "louisiana", "maine", "maryland",
    "massachusetts", "michigan", "minnesota", "mississippi",
    "missouri", "montana", "nebraska", "nevada",
    "new hampshire", "new jersey", "new mexico", "new york",
    "north carolina", "north dakota", "ohio", "oklahoma", "oregon",
    "pennsylvania", "rhode island", "south carolina", "south dakota",
    "tennessee", "texas", "utah", "vermont", "virginia",
    "washington", "west virginia", "wisconsin", "wyoming"
}

INDIA_STATES = {
    "andhra pradesh", "arunachal pradesh", "assam", "bihar",
    "chhattisgarh", "goa", "gujarat", "haryana", "himachal pradesh",
    "jharkhand", "karnataka", "kerala", "madhya pradesh",
    "maharashtra", "manipur", "meghalaya", "mizoram", "nagaland",
    "odisha", "punjab", "rajasthan", "sikkim", "tamil nadu",
    "telangana", "tripura", "uttar pradesh", "uttarakhand",
    "west bengal","delhi"
}

STATE_NAMES = US_STATES | INDIA_STATES


# ------------------------------------------------------------
# Separate address transformations
# ------------------------------------------------------------

state_candidates = []
abbreviation_candidates = []

for old_token, counter in address_transitions.items():

    total = sum(counter.values())

    new_token, count = counter.most_common(1)[0]

    confidence = count / total

    if count < 20:
        continue

    candidate = (
        old_token,
        new_token,
        count,
        total,
        confidence
    )

    # If either side is a state name, classify separately
    if (
        old_token in STATE_NAMES
        or new_token in STATE_NAMES
    ):
        state_candidates.append(candidate)
    else:
        abbreviation_candidates.append(candidate)


# Sort
state_candidates.sort(key=lambda x: (x[4], x[2]), reverse=True)
abbreviation_candidates.sort(key=lambda x: (x[4], x[2]), reverse=True)


# ============================================================
# STATE MAPPINGS
# ============================================================

print("=" * 90)
print("STATE / GEOGRAPHIC MAPPINGS")
print("=" * 90)

for old, new, count, total, confidence in state_candidates[:50]:

    print(
        f"{old:20} -> "
        f"{new:20} | "
        f"count={count:5} | "
        f"total={total:5} | "
        f"confidence={confidence:.3f}"
    )


# ============================================================
# ACTUAL ADDRESS ABBREVIATION CANDIDATES
# ============================================================

print("\n" + "=" * 90)
print("NON-GEOGRAPHIC ADDRESS CANDIDATES")
print("=" * 90)

for old, new, count, total, confidence in abbreviation_candidates[:50]:

    print(
        f"{old:20} -> "
        f"{new:20} | "
        f"count={count:5} | "
        f"total={total:5} | "
        f"confidence={confidence:.3f}"
    )
```

# complete data abbrevation detection

```python

# ============================================================
# STEP 7
# FULL 7.6M-PAIR TOKEN TRANSITION STATISTICS
# ============================================================

# We normalize only for LEARNING the dictionary.
#
# Rules:
#   1. lowercase
#   2. NFKC
#   3. remove periods so L.L.C. can become LLC-like
#   4. turn remaining punctuation into spaces
#   5. split into tokens
#
# We only compare records having the same number of tokens.

con.execute("DROP TABLE IF EXISTS token_transition_stats")


query = """
CREATE TEMP TABLE token_transition_stats AS

WITH comparable AS (

    SELECT
        s1_name,
        matched_name,
        s1_address,
        matched_address,

        -- Remove dots first.
        -- This allows L.L.C. to be treated as LLC-like text.
        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(s1_name),
                        '\\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS name_a,

        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(matched_name),
                        '\\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS name_b,

        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(s1_address),
                        '\\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS addr_a,

        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(matched_address),
                        '\\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS addr_b

    FROM matched_records

    WHERE
        s1_name IS NOT NULL
        AND matched_name IS NOT NULL
        AND s1_address IS NOT NULL
        AND matched_address IS NOT NULL
),

name_aligned AS (

    SELECT
        'business_name' AS field,
        z[1] AS old_token,
        z[2] AS new_token
    FROM comparable,
    UNNEST(list_zip(name_a, name_b)) AS t(z)
    WHERE
        length(name_a) = length(name_b)
),

address_aligned AS (

    SELECT
        'business_address' AS field,
        z[1] AS old_token,
        z[2] AS new_token
    FROM comparable,
    UNNEST(list_zip(addr_a, addr_b)) AS t(z)
    WHERE
        length(addr_a) = length(addr_b)
)

SELECT
    field,
    old_token,
    new_token,
    COUNT(*) AS count
FROM (
    SELECT * FROM name_aligned
    UNION ALL
    SELECT * FROM address_aligned
)
WHERE
    old_token IS NOT NULL
    AND new_token IS NOT NULL
GROUP BY
    field,
    old_token,
    new_token
"""

print("Running full transition analysis...")
con.execute(query)

print("Done.")
```

```python
# ============================================================
# STEP 8
# CALCULATE FULL-DATA TOKEN TRANSITION CONFIDENCE
# ============================================================

confidence_df = con.execute("""
    WITH totals AS (
        SELECT
            field,
            old_token,
            SUM(count) AS total_occurrences
        FROM token_transition_stats
        GROUP BY
            field,
            old_token
    ),

    ranked AS (
        SELECT
            t.field,
            t.old_token,
            t.new_token,
            t.count,
            totals.total_occurrences,

            t.count * 1.0
                / totals.total_occurrences AS confidence,

            ROW_NUMBER() OVER (
                PARTITION BY t.field, t.old_token
                ORDER BY t.count DESC
            ) AS rn

        FROM token_transition_stats t

        JOIN totals
            ON t.field = totals.field
            AND t.old_token = totals.old_token

        WHERE
            t.old_token <> t.new_token
    )

    SELECT
        field,
        old_token,
        new_token,
        count,
        total_occurrences,
        confidence

    FROM ranked

    WHERE rn = 1

    ORDER BY
        field,
        confidence DESC,
        count DESC
""").fetchdf()

print("Candidates:", len(confidence_df))

display(
    confidence_df[
        confidence_df["count"] >= 100
    ].head(100)
)
```

```python
# ============================================================
# STEP 9
# CONDITIONAL TRANSFORMATION CONFIDENCE
#
# Confidence asks:
#
# "When old_token changes, how often does it become new_token?"
#
# We EXCLUDE old_token -> old_token from the denominator.
# ============================================================

confidence_df = con.execute("""
    WITH changed AS (

        SELECT
            field,
            old_token,
            new_token,
            count,

            SUM(count) OVER (
                PARTITION BY field, old_token
            ) AS all_transitions

        FROM token_transition_stats

        WHERE old_token <> new_token
    ),

    ranked AS (

        SELECT
            field,
            old_token,
            new_token,
            count,
            all_transitions,

            count * 1.0
                / all_transitions AS confidence,

            ROW_NUMBER() OVER (
                PARTITION BY field, old_token
                ORDER BY count DESC
            ) AS rn

        FROM changed
    )

    SELECT
        field,
        old_token,
        new_token,
        count,
        all_transitions,
        confidence

    FROM ranked

    WHERE rn = 1

    ORDER BY
        field,
        confidence DESC,
        count DESC
""").fetchdf()


print("Candidates:", len(confidence_df))

display(
    confidence_df[
        confidence_df["count"] >= 100
    ].head(100)
)
```

```python
# ============================================================
# STEP 9 — FULL DATA, EXACTLY ONE TOKEN DIFFERENCE
# ============================================================

con.execute("DROP TABLE IF EXISTS one_change_transitions")

query = r"""
CREATE TEMP TABLE one_change_transitions AS

WITH normalized AS (

    SELECT

        -- ==================================================
        -- BUSINESS NAME — SOURCE 1
        -- ==================================================

        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(coalesce(s1_name, '')),
                        '\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS name_a,


        -- ==================================================
        -- BUSINESS NAME — MATCHED SOURCE
        -- ==================================================

        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(coalesce(matched_name, '')),
                        '\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS name_b,


        -- ==================================================
        -- ADDRESS — SOURCE 1
        -- ==================================================

        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(coalesce(s1_address, '')),
                        '\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS addr_a,


        -- ==================================================
        -- ADDRESS — MATCHED SOURCE
        -- ==================================================

        regexp_split_to_array(
            trim(
                regexp_replace(
                    regexp_replace(
                        lower(coalesce(matched_address, '')),
                        '\.',
                        '',
                        'g'
                    ),
                    '[^[:alnum:]]+',
                    ' ',
                    'g'
                )
            ),
            ' '
        ) AS addr_b

    FROM matched_records
),


-- ==========================================================
-- BUSINESS NAME
-- Keep ONLY pairs with exactly ONE changed token
-- ==========================================================

name_pairs AS (

    SELECT
        'business_name' AS field,
        z[1] AS old_token,
        z[2] AS new_token

    FROM normalized,

    UNNEST(
        list_filter(
            list_zip(name_a, name_b),
            lambda x: x[1] <> x[2]
        )
    ) AS t(z)

    WHERE
        length(name_a) = length(name_b)

        AND
        length(
            list_filter(
                list_zip(name_a, name_b),
                lambda x: x[1] <> x[2]
            )
        ) = 1
),


-- ==========================================================
-- BUSINESS ADDRESS
-- Keep ONLY pairs with exactly ONE changed token
-- ==========================================================

address_pairs AS (

    SELECT
        'business_address' AS field,
        z[1] AS old_token,
        z[2] AS new_token

    FROM normalized,

    UNNEST(
        list_filter(
            list_zip(addr_a, addr_b),
            lambda x: x[1] <> x[2]
        )
    ) AS t(z)

    WHERE
        length(addr_a) = length(addr_b)

        AND
        length(
            list_filter(
                list_zip(addr_a, addr_b),
                lambda x: x[1] <> x[2]
            )
        ) = 1
)


-- ==========================================================
-- COMBINE
-- ==========================================================

SELECT
    field,
    old_token,
    new_token,
    COUNT(*) AS count

FROM (
    SELECT * FROM name_pairs

    UNION ALL

    SELECT * FROM address_pairs
)

WHERE
    old_token <> new_token

GROUP BY
    field,
    old_token,
    new_token
"""

print("Running exact-one-token transition analysis...")

con.execute(query)

print("DONE")
```

```python
# ============================================================
# STEP 13 — CORRECT CANONICAL PAIR CONFIDENCE
# ============================================================
#
# We treat:
#
#     short <-> long
#
# as one representation pair.
#
# Confidence:
#
#     max(direction counts)
#     ---------------------
#     both direction counts
#
# Then we choose:
#
#     short -> long
#
# as the preprocessing mapping.
# ============================================================

pairs = []

for field in ["business_name", "business_address"]:

    field_df = one_change_transitions[
        one_change_transitions["field"] == field
    ]

    pair_counts = {}

    for row in field_df.itertuples(index=False):

        a = str(row.old_token)
        b = str(row.new_token)
        count = int(row.count)

        if a == b:
            continue

        if a in FAKE_NULLS or b in FAKE_NULLS:
            continue

        key = tuple(sorted([a, b]))

        if key not in pair_counts:
            pair_counts[key] = {
                "a_to_b": 0,
                "b_to_a": 0
            }

        if a == key[0]:
            pair_counts[key]["a_to_b"] += count
        else:
            pair_counts[key]["b_to_a"] += count


    # --------------------------------------------------------
    # Convert each pair into short -> canonical
    # --------------------------------------------------------

    for (a, b), counts in pair_counts.items():

        if len(a) == len(b):
            continue

        if len(a) < len(b):
            short = a
            canonical = b

            short_to_long = counts["a_to_b"]
            long_to_short = counts["b_to_a"]

        else:
            short = b
            canonical = a

            short_to_long = counts["b_to_a"]
            long_to_short = counts["a_to_b"]


        total = short_to_long + long_to_short

        if total == 0:
            continue

        pair_confidence = max(
            short_to_long,
            long_to_short
        ) / total

        pairs.append({
            "field": field,
            "short_token": short,
            "canonical_token": canonical,
            "short_to_long_count": short_to_long,
            "long_to_short_count": long_to_short,
            "total": total,
            "pair_confidence": pair_confidence
        })


canonical_pairs = pd.DataFrame(pairs)

canonical_pairs = canonical_pairs.sort_values(
    ["pair_confidence", "total"],
    ascending=[False, False]
).reset_index(drop=True)


print("Total canonical pairs:", len(canonical_pairs))

display(
    canonical_pairs.head(100)
)
```

```python
# ============================================================
# STEP 14 — CLASSIFY LEARNED TOKEN PAIRS
# ============================================================

# Geographic/state tokens discovered from the data.
# These are kept separately from abbreviation mappings.

US_STATES = {
    "alabama", "alaska", "arizona", "arkansas", "california",
    "colorado", "connecticut", "delaware", "florida", "georgia",
    "hawaii", "idaho", "illinois", "indiana", "iowa",
    "kansas", "kentucky", "louisiana", "maine", "maryland",
    "massachusetts", "michigan", "minnesota", "mississippi",
    "missouri", "montana", "nebraska", "nevada",
    "newhampshire", "newjersey", "newmexico", "newyork",
    "northcarolina", "northdakota", "ohio", "oklahoma", "oregon",
    "pennsylvania", "rhodeisland", "southcarolina", "southdakota",
    "tennessee", "texas", "utah", "vermont", "virginia",
    "washington", "westvirginia", "wisconsin", "wyoming"
}

INDIA_STATES = {
    "andhrapradesh", "arunachalpradesh", "assam", "bihar",
    "chhattisgarh", "goa", "gujarat", "haryana", "himachalpradesh",
    "jharkhand", "karnataka", "kerala", "madhyapradesh",
    "maharashtra", "manipur", "meghalaya", "mizoram", "nagaland",
    "odisha", "punjab", "rajasthan", "sikkim", "tamilnadu",
    "telangana", "tripura", "uttarpradesh", "uttarakhand",
    "westbengal"
}

GEOGRAPHIC_WORDS = US_STATES | INDIA_STATES

FAKE_NULLS = {
    "",
    "null",
    "none",
    "na",
    "n/a",
    "nan",
    "nil",
    "unknown",
    "notavailable"
}


# ------------------------------------------------------------
# Classify each learned pair
# ------------------------------------------------------------

def classify_pair(row):

    short = row["short_token"]
    canonical = row["canonical_token"]

    # Fake/null representation
    if short in FAKE_NULLS or canonical in FAKE_NULLS:
        return "fake_null"

    # Geographic/state representation
    if (
        short in GEOGRAPHIC_WORDS
        or canonical in GEOGRAPHIC_WORDS
    ):
        return "geographic"

    # Otherwise it remains a generic candidate
    return "candidate"


canonical_pairs["category"] = canonical_pairs.apply(
    classify_pair,
    axis=1
)


print("CATEGORY COUNTS")
print("=" * 60)

print(
    canonical_pairs["category"]
    .value_counts()
)
```

```python
# ============================================================
# STEP 14B — SHOW NON-GEOGRAPHIC CANDIDATES
# ============================================================

candidates = canonical_pairs[
    (canonical_pairs["category"] == "candidate") &
    (canonical_pairs["total"] >= 100)
].copy()

print("Non-geographic candidates:", len(candidates))

display(
    candidates.sort_values(
        ["pair_confidence", "total"],
        ascending=[False, False]
    ).head(200)
)
```

```python
# ============================================================
# STEP 15 — ABBREVIATION STRUCTURE FILTER
# ============================================================
#
# A candidate is abbreviation-like if the characters of the
# short token appear in the same order inside the canonical
# token.
#
# Examples:
#
#     st   -> street       YES
#     rd   -> road         YES
#     dr   -> drive        YES
#     ave  -> avenue       YES
#     ltd  -> limited      YES
#     inc  -> incorporated YES
#
#     howrah -> kolkata    NO
#     trust  -> center     NO
#     india  -> services   NO
#
# ============================================================

def is_subsequence(short, canonical):
    """
    Returns True if every character of short appears in
    canonical in the same order.
    """

    short = str(short)
    canonical = str(canonical)

    if len(short) >= len(canonical):
        return False

    i = 0

    for char in canonical:

        if i < len(short) and short[i] == char:
            i += 1

    return i == len(short)


# ------------------------------------------------------------
# Apply structural test
# ------------------------------------------------------------

candidates["short_is_subsequence"] = candidates.apply(
    lambda row: is_subsequence(
        row["short_token"],
        row["canonical_token"]
    ),
    axis=1
)


# ------------------------------------------------------------
# Apply additional basic conditions
# ------------------------------------------------------------

abbreviation_candidates = candidates[
    (candidates["short_is_subsequence"]) &
    (candidates["short_token"].str.len() >= 2) &
    (candidates["canonical_token"].str.len() > candidates["short_token"].str.len()) &
    (candidates["total"] >= 100) &
    (candidates["pair_confidence"] >= 0.80)
].copy()


print(
    "Structural abbreviation candidates:",
    len(abbreviation_candidates)
)


display(
    abbreviation_candidates
    .sort_values(
        ["field", "pair_confidence", "total"],
        ascending=[True, False, False]
    )
    .head(200)
)
```

```python
# ============================================================
# STEP 17 — CLASSIFY LEARNED ADDRESS VARIANTS
# ============================================================

# Standard street/address suffixes.
#
# These are NOT being learned manually.
# They are only used to identify what category a learned
# transformation belongs to.

ADDRESS_SUFFIXES = {
    "street",
    "road",
    "drive",
    "avenue",
    "lane",
    "court",
    "circle",
    "place",
    "boulevard",
    "trail",
    "highway",
    "terrace",
    "parkway",
    "cove",
    "square",
    "ridge",
    "turnpike",
    "point",
    "fort",
    "mount",
    "way",
    "alley",
    "plaza",
    "expressway",
    "junction",
    "loop",
    "route",
    "highway"
}


# ------------------------------------------------------------
# Classify every approved structural candidate
# ------------------------------------------------------------

approved = abbreviation_candidates[
    (abbreviation_candidates["total"] >= 100) &
    (abbreviation_candidates["pair_confidence"] >= 0.80) &
    (abbreviation_candidates["short_is_subsequence"])
].copy()


def classify_address_variant(row):

    canonical = row["canonical_token"]

    # -----------------------------------------------
    # Standard address suffix
    # -----------------------------------------------

    if (
        row["field"] == "business_address"
        and canonical in ADDRESS_SUFFIXES
    ):
        return "address_abbreviation"

    # -----------------------------------------------
    # Everything else for now
    # -----------------------------------------------

    return "other"


approved["variant_type"] = approved.apply(
    classify_address_variant,
    axis=1
)


print("VARIANT TYPE COUNTS")
print("=" * 60)

print(
    approved["variant_type"]
    .value_counts()
)
```

```python
# ============================================================
# STEP 18 — BUSINESS NAME ABBREVIATION STRUCTURE
# ============================================================

name_candidates = approved[
    approved["field"] == "business_name"
].copy()


# ------------------------------------------------------------
# Character-length features
# ------------------------------------------------------------

name_candidates["short_len"] = (
    name_candidates["short_token"].str.len()
)

name_candidates["canonical_len"] = (
    name_candidates["canonical_token"].str.len()
)

name_candidates["length_gap"] = (
    name_candidates["canonical_len"]
    - name_candidates["short_len"]
)

name_candidates["compression_ratio"] = (
    name_candidates["short_len"]
    / name_candidates["canonical_len"]
)


# ------------------------------------------------------------
# Initial abbreviation rule
#
# We want:
#
#   short token
#   AND
#   at least 2 characters shorter
#   AND
#   strong pair confidence
#
# This rejects most one-character typo variants.
# ------------------------------------------------------------

name_abbreviation_candidates = name_candidates[
    (name_candidates["short_len"] <= 4) &
    (name_candidates["length_gap"] >= 2) &
    (name_candidates["pair_confidence"] >= 0.80) &
    (name_candidates["total"] >= 100) &
    (name_candidates["short_is_subsequence"])
].copy()


print(
    "Business-name abbreviation candidates:",
    len(name_abbreviation_candidates)
)


display(
    name_abbreviation_candidates[
        [
            "short_token",
            "canonical_token",
            "short_len",
            "canonical_len",
            "total",
            "pair_confidence"
        ]
    ]
    .sort_values(
        ["pair_confidence", "total"],
        ascending=[False, False]
    )
)
```

```python
# ============================================================
# STEP 21A — LEARN TYPO / SPELLING VARIANTS
# ============================================================

from rapidfuzz.distance import Levenshtein


# ------------------------------------------------------------
# Start from ALL canonical pairs.
#
# We deliberately do NOT start only from abbreviation_candidates
# because typo variants were previously filtered out.
# ------------------------------------------------------------

typo_candidates = canonical_pairs.copy()


# ------------------------------------------------------------
# Calculate edit distance
# ------------------------------------------------------------

typo_candidates["edit_distance"] = typo_candidates.apply(
    lambda row: Levenshtein.distance(
        str(row["short_token"]),
        str(row["canonical_token"])
    ),
    axis=1
)


# ------------------------------------------------------------
# Length ratio
# ------------------------------------------------------------

typo_candidates["length_ratio"] = (
    typo_candidates["short_token"].str.len()
    /
    typo_candidates["canonical_token"].str.len()
)


# ------------------------------------------------------------
# TYPO RULE
#
# We want:
#
# edit distance <= 2
# confidence >= 0.80
# at least 100 observations
# canonical form longer than variant
#
# ------------------------------------------------------------

typo_candidates = typo_candidates[
    (typo_candidates["edit_distance"] <= 2) &
    (typo_candidates["pair_confidence"] >= 0.80) &
    (typo_candidates["total"] >= 100) &
    (
        typo_candidates["canonical_token"].str.len()
        >=
        typo_candidates["short_token"].str.len()
    )
].copy()


print(
    "Potential typo candidates:",
    len(typo_candidates)
)


display(
    typo_candidates[
        [
            "field",
            "short_token",
            "canonical_token",
            "edit_distance",
            "total",
            "pair_confidence"
        ]
    ]
    .sort_values(
        ["edit_distance", "pair_confidence", "total"],
        ascending=[True, False, False]
    )
    .head(200)
)
```

```python
# ============================================================
# STEP 21B — CHECK KNOWN TYPO EXAMPLES
# ============================================================

known_typos = [
    ("stret", "street"),
    ("drve", "drive"),
    ("drie", "drive"),
    ("dive", "drive"),
    ("steet", "street"),
    ("aenue", "avenue"),
    ("rad", "road"),
    ("rod", "road"),
    ("lae", "lane"),
    ("foods", "foodworks"),
    ("ssociates", "associates"),
    ("privte", "private"),
]


for short, canonical in known_typos:

    result = typo_candidates[
        (typo_candidates["short_token"] == short) &
        (typo_candidates["canonical_token"] == canonical)
    ]

    print(f"\n{short} -> {canonical}")

    if len(result):
        display(
            result[
                [
                    "field",
                    "short_token",
                    "canonical_token",
                    "edit_distance",
                    "total",
                    "pair_confidence"
                ]
            ]
        )
    else:
        print("Not selected as typo candidate")
```

```python
# ============================================================
# STEP 21C — STRICT TYPO / SPELLING DICTIONARY
# ============================================================

# Start from the candidates discovered in Step 21A.
# We intentionally make this conservative.

final_typo_candidates = typo_candidates[
    # Only one-character edits
    (typo_candidates["edit_distance"] == 1) &

    # Strong evidence from matched entities
    (typo_candidates["pair_confidence"] >= 0.80) &
    (typo_candidates["total"] >= 100) &

    # Avoid tiny tokens such as:
    # nc -> inc
    # nd -> and
    # pc -> inc
    (typo_candidates["short_token"].str.len() >= 4) &

    # The typo should be reasonably close in length
    (
        typo_candidates["length_ratio"] >= 0.80
    ) &

    # Canonical spelling should not be shorter
    (
        typo_candidates["canonical_token"].str.len()
        >=
        typo_candidates["short_token"].str.len()
    )
].copy()


# ------------------------------------------------------------
# Remove geographic tokens
# ------------------------------------------------------------

final_typo_candidates = final_typo_candidates[
    ~(
        final_typo_candidates["short_token"].isin(
            GEOGRAPHIC_WORDS
        )
        |
        final_typo_candidates["canonical_token"].isin(
            GEOGRAPHIC_WORDS
        )
    )
].copy()


# ------------------------------------------------------------
# Remove anything already classified as an abbreviation
# ------------------------------------------------------------

existing_abbreviation_pairs = set()

for short, canonical in business_name_abbreviations.items():
    existing_abbreviation_pairs.add(
        ("business_name", short, canonical)
    )

for short, canonical in business_address_abbreviations.items():
    existing_abbreviation_pairs.add(
        ("business_address", short, canonical)
    )


final_typo_candidates = final_typo_candidates[
    ~final_typo_candidates.apply(
        lambda row:
            (
                row["field"],
                row["short_token"],
                row["canonical_token"]
            )
            in existing_abbreviation_pairs,
        axis=1
    )
].copy()


# ------------------------------------------------------------
# Final display
# ------------------------------------------------------------

print(
    "FINAL TYPO CANDIDATES:",
    len(final_typo_candidates)
)

display(
    final_typo_candidates[
        [
            "field",
            "short_token",
            "canonical_token",
            "edit_distance",
            "length_ratio",
            "total",
            "pair_confidence"
        ]
    ]
    .sort_values(
        ["field", "total"],
        ascending=[True, False]
    )
)
```

```python
# ============================================================
# STEP 21D — TYPO SANITY CHECK
# ============================================================

known_cases = [
    ("stret", "street"),
    ("drve", "drive"),
    ("drie", "drive"),
    ("dive", "drive"),
    ("steet", "street"),
    ("aenue", "avenue"),
    ("rad", "road"),
    ("rod", "road"),
    ("lae", "lane"),

    # These should NOT be accepted
    ("nc", "inc"),
    ("nd", "and"),
    ("pc", "inc"),
    ("pc", "llc"),
    ("kerala", "keralam"),
]


for short, canonical in known_cases:

    result = final_typo_candidates[
        (final_typo_candidates["short_token"] == short) &
        (final_typo_candidates["canonical_token"] == canonical)
    ]

    status = "KEEP" if len(result) else "REJECT"

    print(
        f"{short:10} -> {canonical:12} {status}"
    )
```

```python
# ============================================================
# STEP 21E — PRECISION-FIRST TYPO CANDIDATES
# ============================================================
#
# IMPORTANT:
# These mappings are NOT applied to preprocessing yet.
#
# F0.5 favors precision, so we only identify candidates first.
# ============================================================

# Start from typo_candidates generated in STEP 21A.

tier2_typo_candidates = typo_candidates[
    # Exactly one character difference
    (typo_candidates["edit_distance"] == 1) &

    # Strong evidence from matched entities
    (typo_candidates["pair_confidence"] >= 0.95) &

    # Enough observations
    (typo_candidates["total"] >= 100) &

    # Avoid extremely short tokens
    (typo_candidates["short_token"].str.len() >= 4) &

    # Similar length — characteristic of a typo,
    # not a normal abbreviation
    (typo_candidates["length_ratio"] >= 0.75) &

    # Canonical spelling should not be shorter
    (
        typo_candidates["canonical_token"].str.len()
        >=
        typo_candidates["short_token"].str.len()
    )
].copy()


# ------------------------------------------------------------
# Remove geographic transformations
# ------------------------------------------------------------

tier2_typo_candidates = tier2_typo_candidates[
    ~(
        tier2_typo_candidates["short_token"].isin(GEOGRAPHIC_WORDS)
        |
        tier2_typo_candidates["canonical_token"].isin(GEOGRAPHIC_WORDS)
    )
].copy()


# ------------------------------------------------------------
# Remove mappings already classified as abbreviations
# ------------------------------------------------------------

tier2_typo_candidates = tier2_typo_candidates[
    ~tier2_typo_candidates.apply(
        lambda row:
            (
                row["field"],
                row["short_token"],
                row["canonical_token"]
            )
            in existing_abbreviation_pairs,
        axis=1
    )
].copy()


# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print(
    "Tier-2 typo candidates:",
    len(tier2_typo_candidates)
)

display(
    tier2_typo_candidates[
        [
            "field",
            "short_token",
            "canonical_token",
            "edit_distance",
            "length_ratio",
            "total",
            "pair_confidence"
        ]
    ]
    .sort_values(
        ["pair_confidence", "total"],
        ascending=[False, False]
    )
)
```

```python
# ============================================================
# STEP 22.5 — FINAL SAFE PRODUCTION ABBREVIATIONS
# ============================================================

# ============================================================
# 1. BUSINESS NAME ABBREVIATIONS
# ============================================================

LEGAL_NAME_CANONICALS = {
    "incorporated",
    "limited",
    "corporation",
    "private",
    "company",
    "llc",
    "llp",
    "plc"
}


safe_name_df = name_candidates[
    # Strong evidence
    (name_candidates["total"] >= 100) &
    (name_candidates["pair_confidence"] >= 0.80) &

    # Must actually be shorter
    (
        name_candidates["short_token"].str.len()
        <
        name_candidates["canonical_token"].str.len()
    ) &

    # Subsequence structure
    (name_candidates["short_is_subsequence"]) &

    # Canonical must be a legitimate legal/business term
    (
        name_candidates["canonical_token"].isin(
            LEGAL_NAME_CANONICALS
        )
    ) &

    # IMPORTANT:
    # Require real compression.
    #
    # inc/incorporated = 0.25
    # ltd/limited      = 0.43
    #
    # privte/private   = 0.86  -> rejected
    #
    (
        name_candidates["short_token"].str.len()
        /
        name_candidates["canonical_token"].str.len()
        <= 0.65
    )
].copy()


# ------------------------------------------------------------
# Build name dictionary
# ------------------------------------------------------------

production_name_abbreviations = dict(
    zip(
        safe_name_df["short_token"],
        safe_name_df["canonical_token"]
    )
)


# ============================================================
# 2. BUSINESS ADDRESS ABBREVIATIONS
# ============================================================

# IMPORTANT:
# Use ONLY the address-abbreviation dataframe generated
# before typo candidates were mixed in.

safe_address_df = address_abbreviations[
    (address_abbreviations["total"] >= 100) &
    (address_abbreviations["pair_confidence"] >= 0.80)
].copy()


# ------------------------------------------------------------
# Compression ratio
# ------------------------------------------------------------

safe_address_df["compression_ratio"] = (
    safe_address_df["short_token"].str.len()
    /
    safe_address_df["canonical_token"].str.len()
)


# ------------------------------------------------------------
# Keep genuine abbreviations
#
# rd  -> road       0.50
# dr  -> drive      0.40
# st  -> street     0.33
# ave -> avenue     0.50
# pkwy -> parkway   0.67
#
# Reject:
#
# stret -> street   0.83
# rad   -> road     0.75
# drve  -> drive    0.80
# ------------------------------------------------------------

safe_address_df = safe_address_df[
    safe_address_df["compression_ratio"] <= 0.70
].copy()


# ------------------------------------------------------------
# Build address dictionary
# ------------------------------------------------------------

production_address_abbreviations = dict(
    zip(
        safe_address_df["short_token"],
        safe_address_df["canonical_token"]
    )
)


# ============================================================
# DISPLAY FINAL PRODUCTION RULES
# ============================================================

print("=" * 70)
print("FINAL PRODUCTION BUSINESS NAME ABBREVIATIONS")
print("=" * 70)

for short, canonical in production_name_abbreviations.items():
    print(f"{short:15} -> {canonical}")

print(
    "\nBusiness-name rules:",
    len(production_name_abbreviations)
)


print("\n" + "=" * 70)
print("FINAL PRODUCTION BUSINESS ADDRESS ABBREVIATIONS")
print("=" * 70)

for short, canonical in production_address_abbreviations.items():
    print(f"{short:15} -> {canonical}")

print(
    "\nAddress rules:",
    len(production_address_abbreviations)
)
```

```python
%%writefile -a /kaggle/working/src/preprocess.py

# ============================================================
# Learned abbreviation system
# ============================================================

import json
from pathlib import Path

ABBREVIATION_JSON = (
    Path(__file__).resolve().parent.parent
    / "abbreviations_production.json"
)


def load_abbreviation_rules():
    """
    Load production abbreviation mappings learned from
    the training data.
    """
    with open(ABBREVIATION_JSON, "r", encoding="utf-8") as f:
        return json.load(f)


ABBREVIATION_RULES = load_abbreviation_rules()

BUSINESS_NAME_ABBREVIATIONS = ABBREVIATION_RULES.get(
    "business_name", {}
)

BUSINESS_ADDRESS_ABBREVIATIONS = ABBREVIATION_RULES.get(
    "business_address", {}
)


def apply_abbreviations(text, mapping):
    """
    Apply token-level abbreviation replacements.

    Only complete tokens are replaced, so substrings inside
    normal words are not accidentally modified.
    """
    if not text or not mapping:
        return text

    # Longer tokens first
    tokens = sorted(mapping.keys(), key=len, reverse=True)

    for token in tokens:
        replacement = mapping[token]

        pattern = rf"\b{re.escape(token)}\b"

        text = re.sub(
            pattern,
            replacement,
            text
        )

    return text
```

```python
%%writefile -a /kaggle/working/src/preprocess.py

# ============================================================
# Normalization using learned abbreviation mappings
# ============================================================

def normalize_business_name(text):
    """
    Normalize business names using learned production
    abbreviation mappings.
    """
    text = normalize_unicode(text)

    if not text:
        return ""

    # Lowercase while preserving Unicode
    text = text.lower()

    # Apply learned business-name abbreviations
    text = apply_abbreviations(
        text,
        BUSINESS_NAME_ABBREVIATIONS
    )

    # '&' -> 'and'
    text = text.replace("&", " and ")

    # Remove punctuation while preserving Unicode
    cleaned = []

    for char in text:
        category = unicodedata.category(char)

        if category.startswith("P"):
            cleaned.append(" ")
        else:
            cleaned.append(char)

    text = "".join(cleaned)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


def normalize_business_address(text):
    """
    Normalize business addresses using learned production
    abbreviation mappings.
    """
    text = normalize_unicode(text)

    if not text:
        return ""

    # Lowercase while preserving Unicode
    text = text.lower()

    # Apply learned address abbreviations
    text = apply_abbreviations(
        text,
        BUSINESS_ADDRESS_ABBREVIATIONS
    )

    # '&' -> 'and'
    text = text.replace("&", " and ")

    # Remove punctuation while preserving Unicode
    cleaned = []

    for char in text:
        category = unicodedata.category(char)

        if category.startswith("P"):
            cleaned.append(" ")
        else:
            cleaned.append(char)

    text = "".join(cleaned)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```
